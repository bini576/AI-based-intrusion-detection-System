```python
import joblib
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder

def load_model(model_path):
    """Load the trained Decision Tree model."""
    try:
        model = joblib.load(model_path)
        print("✅ Model loaded successfully!")
        return model
    except Exception as e:
        print(f"❌ Error loading model: {e}")
        exit()

def preprocess_data(data_path, selected_features):
    """Load and preprocess the new test data."""
    try:
        data = pd.read_csv(data_path)
        print("✅ Data loaded successfully!")

        # Encode categorical features
        def le(df):
            for col in df.columns:
                if df[col].dtype == 'object':
                    label_encoder = LabelEncoder()
                    df[col] = label_encoder.fit_transform(df[col])
        le(data)

        # Drop unnecessary columns (like in training)
        if 'num_outbound_cmds' in data.columns:
            data.drop(['num_outbound_cmds'], axis=1, inplace=True)

        # Select the same features used during training
        data = data[selected_features]

        # Standardize the data
        scale = StandardScaler()
        X = scale.fit_transform(data)

        print("✅ Data preprocessing complete!")
        return X
    except Exception as e:
        print(f"❌ Error preprocessing data: {e}")
        exit()

def make_predictions(model, X):
    """Make predictions using the loaded model."""
    try:
        predictions = model.predict(X)
        print("✅ Predictions made successfully!")
        return predictions
    except Exception as e:
        print(f"❌ Error making predictions: {e}")
        exit()

def save_predictions(predictions, output_path):
    """Save predictions to a CSV file."""
    try:
        df = pd.DataFrame(predictions, columns=['Prediction'])
        df.to_csv(output_path, index=False)
        print(f"✅ Predictions saved to {output_path}!")
    except Exception as e:
        print(f"❌ Error saving predictions: {e}")
        exit()

if __name__ == "__main__":
    # Path to your saved model
    model_path = "decision_tree_model.pkl"
    
    # Path to the new test data
    new_data_path = "Test_data.csv"
    
    # Path to save predictions
    output_path = "predictions.csv"

    # Load the model
    model = load_model(model_path)

    # Manually list the selected features (from your training process)
    selected_features = [
        "duration", "src_bytes", "dst_bytes", "wrong_fragment", 
        "count", "srv_count", "same_srv_rate", 
        "diff_srv_rate", "srv_diff_host_rate", "dst_host_count"
    ]

    # Preprocess the new data
    X_new = preprocess_data(new_data_path, selected_features)

    # Make predictions
    predictions = make_predictions(model, X_new)

    # Save predictions to CSV
    save_predictions(predictions, output_path)
```
